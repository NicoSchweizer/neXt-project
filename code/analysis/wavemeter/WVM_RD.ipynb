```python
import ltatools as lta
import numpy as np
import matplotlib.pyplot as plt
import allantools as at
```

```python
df_RD = lta.load_lta_file('data/RD/16.07.2026, 17.09,  267,3349378 THz.lta', cleanup=False)

df_RD = df_RD[(df_RD['time_s'] > 0) & (df_RD['time_s'] < 30)].reset_index(drop=True)

df_RD['time_s'] -= df_RD['time_s'].iloc[0]
df_RD['time_ms'] -= df_RD['time_s'].iloc[0]

lta.plot(data=df_RD, kind='timeseries', lines=True, save=f'fig/RD/timeseries_RD.pdf', relative=True, freq_unit='MHz', figsize=(9,4), tick_direction='in')
lta.plot(data=df_RD, kind='adev', quantity='frequency', errorbars=True, unit='MHz', save=f'fig/RD/adev_RD.pdf', regions=True, tick_direction='in')
```
